```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

pd.set_option('display.max_colwidth', None)
```

```python
data1 = pd.read_csv('train.csv')
data2 = pd.read_csv('test.csv')

train_df = data1.copy()
test_df = data2.copy()

print("First 5 rows train data: \n", train_df.head())
print()
print("First 5 rows test data: \n", test_df.head())
```

```python
map_enc_cols_train = [
    'satisfaction'
]

dummy_cols = [
    'Gender',
    'Customer Type',
    'Type of Travel',
    'Class'
]

for col in map_enc_cols_train:
    train_df[col] = train_df[col].astype(int)

for col in dummy_cols:
    dummies_train = pd.get_dummies(train_df[col], prefix=col, dtype=int)
    dummies_test = pd.get_dummies(test_df[col], prefix=col, dtype=int)

    dummies_test = dummies_test.reindex(columns=dummies_train.columns, fill_value=0)

    train_df = pd.concat([train_df, dummies_train], axis=1)
    test_df = pd.concat([test_df, dummies_test], axis=1)
```

```python
inflight_service_cols = ['Seat comfort', 'Inflight entertainment', 'On-board service', 
                        'Leg room service', 'Baggage handling', 'Checkin service', 
                        'Cleanliness', 'Online boarding']

train_df['entertainment_x_seat'] = train_df['Inflight entertainment'] * train_df['Seat comfort']
test_df['entertainment_x_seat'] = test_df['Inflight entertainment'] * test_df['Seat comfort']

train_df['entertainment_x_boarding'] = train_df['Inflight entertainment'] * train_df['Online boarding']
test_df['entertainment_x_boarding'] = test_df['Inflight entertainment'] * test_df['Online boarding']

train_df['comfort_x_boarding'] = train_df['Seat comfort'] * train_df['Online boarding']
test_df['comfort_x_boarding'] = test_df['Seat comfort'] * test_df['Online boarding']

train_df['service_grade'] = train_df['Seat comfort'] + train_df['Inflight entertainment'] + train_df['On-board service'] + train_df['Leg room service'] + train_df['Baggage handling'] + train_df['Checkin service'] + train_df['Cleanliness'] + train_df['Online boarding']
test_df['service_grade'] = test_df['Seat comfort'] + test_df['Inflight entertainment'] + test_df['On-board service'] + test_df['Leg room service'] + test_df['Baggage handling'] + test_df['Checkin service'] + test_df['Cleanliness'] + test_df['Online boarding']

train_df['entertainment_x_cleanliness'] = train_df['Inflight entertainment'] * train_df['Cleanliness']
test_df['entertainment_x_cleanliness'] = test_df['Inflight entertainment'] * test_df['Cleanliness']

train_df['distance_x_comfort'] = train_df['entertainment_x_cleanliness'] * train_df['Flight Distance']
test_df['distance_x_comfort'] = test_df['entertainment_x_cleanliness'] * test_df['Flight Distance']
```

```python
train_df.to_csv("full_fe_df.csv", index=False)
test_df.to_csv("full_fe_df_test.csv", index=False)
```

```python
train_df = train_df.drop(columns=inflight_service_cols)
test_df = test_df.drop(columns=inflight_service_cols)

train_df.to_csv("simplified_fe_df.csv", index=False)
test_df.to_csv("simplified_fe_df_test.csv", index=False)
```
